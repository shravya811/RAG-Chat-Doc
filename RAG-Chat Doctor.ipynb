```python
QA_chain = ConversationalRetrievalChain.from_llm(
    llm=llm,
    retriever=contexual_retriever,
    memory=memory, \
    return_source_documents=True,      # to return the source documents along with the answer
    verbose=False,
    combine_docs_chain_kwargs={"prompt": prompt},
    output_key='answer'
)
 
print("Welcome to the AI Medical Consultant Chatbot!")
```

```python
def ask_question(question: str) -> dict:
    result = QA_chain.invoke({"question": question})
   
    print("=" * 70)
    print(f"Q: {question}")
    print(f"A: {result['answer']}")
    print("-" * 70)
    print("Sources:")
    for i, doc in enumerate(result["source_documents"], start=1):
        meta   = doc.metadata
        source = meta.get("source", "Unknown").split("\\")[-1]  # filename only
        page   = meta.get("page", "?")
        print(f"  [{i}] {source}  |  Page {page}")
        print(f"       {doc.page_content[:150].strip()}...")
    print("=" * 70)
```

```python
ask_question("I am suffering from severe headaches and nausea. What could be the possible causes and what should I do?")
```
